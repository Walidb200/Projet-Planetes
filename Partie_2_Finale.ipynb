```python
soleil_rayon_ua = 0.004649
ua = 149597887.5

    
def dessine_planetes_3d(k):
    S = sphere(center=(0, 0, 0), size = 30, opacity = 0)
    for planete in systeme_solaire.keys():
        ang = (2*pi / systeme_solaire[planete]["période"])*3*k # Le 3 permet d'accélérer et de devoir produire moins de modélisation
        demi_g_axe = systeme_solaire[planete]["demi_g_axe_ua"]
        demi_p_axe = systeme_solaire[planete]["demi_p_axe_ua"]
        inclinaison =  systeme_solaire[planete]["inclinaison"]*pi/180 # On converti l'inclinaison en radiant
        x =  -systeme_solaire[planete]["c_ua"] + demi_g_axe * cos(ang)
        y = demi_p_axe * sin(ang) * cos(inclinaison)
        z = demi_p_axe * sin(ang) * sin(inclinaison)
        rayon_ua = 1000*systeme_solaire[planete]["rayon_km"]/ua #On multiplie par 1000 les rayon des planetes
        S += sphere((x,y,z), size=rayon_ua, color=systeme_solaire[planete]["couleur"])
        if planete == "Terre":
            ang = (2*pi / satellites["Lune"]["période"])*3*k # Le 3 permet d'accélérer et de devoir produire moins de modélisation
            demi_g_axe = satellites["Lune"]["demi_g_axe_ua"]
            demi_p_axe = satellites["Lune"]["demi_p_axe_ua"]
            inclinaison =  satellites["Lune"]["inclinaison"]*pi/180 # On converti l'inclinaison en radiant
            #Coordonnées de la Lune par rapport à la Terre
            xl_t =  -satellites["Lune"]["c_ua"] + demi_g_axe * cos(ang) 
            yl_t = demi_p_axe * sin(ang) * cos(inclinaison)
            zl_t = demi_p_axe * sin(ang) * sin(inclinaison)
            #Coordonnées de la Lune par rapport au soleil (origine)
            xl = 23*xl_t + x
            yl = 23*yl_t + y
            zl = 23*zl_t + z
            rayon_ua = 1000*satellites["Lune"]["rayon_km"]/ua #On multiplie par 1000 le rayon de la Lune
            S += sphere((xl,yl,zl), size=rayon_ua, color=satellites["Lune"]["couleur"])
            
    return S

    
def animation_systeme_solaire(N):
    # On modélise une seule fois les ellipses et le Soleil
    S = sphere(center=(0, 0, 0), size = 30, opacity = 0)
    S += sphere(center=(0, 0, 0), size = soleil_rayon_ua*10, opacity = 100, color="orange") # On multiplie par 10 le rayon du soleil pour le voir

    for planete in systeme_solaire.keys(): 
        demi_g_axe = systeme_solaire[planete]["demi_g_axe_ua"]
        demi_p_axe = systeme_solaire[planete]["demi_p_axe_ua"]
        inclinaison_rad =  systeme_solaire[planete]["inclinaison"]*pi/180 # On converti l'inclinaison en radiant
        t = var("t")
        x = -systeme_solaire[planete]["c_ua"] + demi_g_axe * cos(t)
        y = demi_p_axe * sin(t) * cos(inclinaison_rad)
        z = demi_p_axe * sin(t) * sin(inclinaison_rad)
        S += parametric_plot3d((x, y, z), (t, 0, 2*pi), color="black") 
    liste = [ S + dessine_planetes_3d(k) for k in range(1, N)]    # On ajoute les planetes bien positionnées
    
    return animate(liste, figsize=(25, 25))
anim = animation_systeme_solaire(300)
```
